#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.
- Show the outputted table.

</div>

```r
output_table <- women_march_tidy |>
         select(seed, reg_wins_pct) |>
         pivot_longer(seed:reg_wins_pct, 
         names_to="variable",
         values_to="value") |>
         group_by(variable) |>
         summarise(
         "mean"=mean(value, na.rm=TRUE),
         "max"=max(value, na.rm=TRUE),
         "min"=min(value, na.rm=TRUE))

output_table                      
```

Visualization 1: The histogram compares the distribution of regular season win percentage for teams that recieved automatic bids and at-large bids.It is relevant because win percentage is one of our 2 predictors of bid type.
Most at-large teams won somewhere between 65% and 80% of their regular season games. Automatic bid teams are much more spread out; They make up most of the teams at the very top(above 85%) and at the very bottom(below 55%). But in the middle, from about 60% to 85% , the two groups overlap a lot. So win percentage on its own doesn't tell us much about which type of bid a team got.

```r
options(repr.plot.width = 8, repr.plot.height = 5)
Visualization_1<- ggplot(women_march_tidy, aes(x=reg_wins_pct, fill=bid))+
                  geom_histogram(binwidth=5,position="identity",alpha=0.4)+
                  labs(title="Distribution of regular season win percentage by bid type",
                       x="Regular season win percentage(%)",
                       y="Number of teams",
                       fill="Bid type")
Visualization_1               
```

Visualization_2 :This bar chart shows the proportion of teams with automatic bids at each seed. It's relevant because seed is our second predictor of bid type. The pattern is U-shaped;Nearly all teams seeded 13-16 got automatic bids, and so did more than half of the 1 and 2 seeds, likely because many top teams also win their conference titles. Seeds 3 to 10, on the other hand, are mostly at-large teams. So seed is strong predictor for low seeds , but less clear-cut for the rest.

```r
Visualization_2<- women_march_tidy |>
                  group_by(seed) |>
                  summarise(prop_auto=mean(bid=="auto")) |>
                  ggplot(aes(x=seed, y=prop_auto)) +
                  geom_bar(stat="identity") +
                  labs(x="tournament seed",
                       y="proportion of automatic bids",
                       title="proportion of automatic by seed")
Visualization_2
```
